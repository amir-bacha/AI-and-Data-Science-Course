*Indexing and Slicing of Array*

*Indexing of One dimensional array*


```python
import numpy as np
array=np.array([1,2,3,4,5,6])
print(array)
```

```python
# indexing meas accessing all the elements of array 
# For example we have an array arr=[1, 2, 3, 4, 5] here the 1 is stored at index 0, 2 at index 1, 3 at index 2 ...
print(array[0]) # 1
print(array[1]) # 2
print(array[4]) # 5
# and if i write array[6] it will cause an error because its out of indexes
```

```python
# loop to access all element of an array
for arr in array:
    print(arr)
```

*Slicing of One Dimensional array*


```python
# Slicing means getting the specific portion from an array 
# syntax is array[start:stop:step]  Note: start is inclusive while the stop is exclusive
nums=np.array([20,30,40,50,60])
Slicednums=nums[0:2]
print(Slicednums)
print(nums)
```

```python
Slicednums=nums[0: :2] # 20,40,60
print(Slicednums)
```

```python
Slicednums=nums[-1]
print(Slicednums)
```

*indexing of two dimensional array in numpy*

```python
import numpy as np
matrix=np.array([[1,2,3,4],[5,6,7,8]])
# to traverse all the element of an array
for i in range(matrix.shape[0]):
    for j in range(matrix.shape[1]):
        print(matrix[i,j],end=" ")
    print()
```

```python
# Slicing in Double dimensional array
# Slicing means cutting a piece of from a thing
matrix=np.array(
    [[20,30,40,50],
    [60,70,80,90],
    [100,110,120,130]]
    )


# Syntax matrix[start,end]
print(matrix[0:2,1:4])
print(matrix[0:2,1:4:2])
```
